```python
import datetime

# --- DADOS MOCKADOS (Simulando o Banco de Dados) ---
funcionarios = {
    1: {"nome": "João Silva", "matricula": "F1001", "setor": "Soldagem"},
    2: {"nome": "Maria Souza", "matricula": "F1002", "setor": "Montagem"}
}

epis = {
    101: {"equipamento": "Máscara de Solda", "validade": "2026-12-01", "status": "Disponível"},
    102: {"equipamento": "Luva de Raspa", "validade": "2026-10-15", "status": "Disponível"},
    103: {"equipamento": "Capacete", "validade": "2026-09-10", "status": "Disponível"} # Perto do vencimento
}

registros_uso = []

# --- FUNÇÕES DO SISTEMA ---
def limpar_tela():
    print("\n" * 2)
    print("-" * 40)

def alertar_vencimentos():
    hoje = datetime.date.today()
    print("\n⚠️ ALERTAS DE VENCIMENTO ⚠️")
    alertas = 0
    for id_epi, dados in epis.items():
        ano, mes, dia = map(int, dados['validade'].split('-'))
        data_val = datetime.date(ano, mes, dia)
        dias_restantes = (data_val - hoje).days

        if dias_restantes <= 30:
            print(f"-> O EPI {dados['equipamento']} (ID: {id_epi}) vence em {dias_restantes} dias!")
            alertas += 1
    if alertas == 0:
        print("Nenhum EPI próximo do vencimento.")

def registrar_entrega():
    print("\n--- REGISTRAR ENTREGA DE EPI ---")
    try:
        id_func = int(input("Digite o ID do Funcionário: "))
        if id_func not in funcionarios:
            print("❌ Funcionário não encontrado.")
            return

        id_epi = int(input("Digite o ID do EPI: "))
        if id_epi not in epis:
            print("❌ EPI não encontrado.")
            return

        data_hoje = datetime.date.today().strftime("%Y-%m-%d")

        novo_registro = {
            "id_registro": len(registros_uso) + 1,
            "funcionario": funcionarios[id_func]['nome'],
            "epi": epis[id_epi]['equipamento'],
            "data_entrega": data_hoje,
            "conformidade": True
        }

        registros_uso.append(novo_registro)
        epis[id_epi]['status'] = "Em Uso"
        print(f"\n✅ SUCESSO: {epis[id_epi]['equipamento']} entregue para {funcionarios[id_func]['nome']}!")

    except ValueError:
        print("❌ Erro: Por favor, digite apenas números para os IDs.")

def exibir_relatorio():
    print("\n--- RELATÓRIO DE ENTREGAS ---")
    if not registros_uso:
        print("Nenhum registro efetuado ainda.")
    else:
        for reg in registros_uso:
            print(f"Registro #{reg['id_registro']} | Func: {reg['funcionario']} | EPI: {reg['epi']} | Data: {reg['data_entrega']}")

# --- FLUXO PRINCIPAL (Telas Navegáveis) ---
def iniciar_sistema():
    print("🛡️ Bem-vindo ao VisioGuard 🛡️")
    print("Iniciando sistema...")
    alertar_vencimentos()

    while True:
        limpar_tela()
        print("MENU PRINCIPAL")
        print("1. Registrar Entrega de EPI")
        print("2. Ver Relatório de Entregas")
        print("3. Sair")

        opcao = input("Escolha uma opção: ")

        if opcao == '1':
            registrar_entrega()
            input("\nPressione ENTER para voltar ao menu...")
        elif opcao == '2':
            exibir_relatorio()
            input("\nPressione ENTER para voltar ao menu...")
        elif opcao == '3':
            print("Encerrando VisioGuard. Segurança em primeiro lugar!")
            break
        else:
            print("❌ Opção inválida.")

# Rodar o protótipo
iniciar_sistema()
```
